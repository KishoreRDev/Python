## 6.Apply

```python
import pandas as pd
insurance_data = pd.read_csv("Insurance.csv")
insurance_data.describe()
```

### 6.1 Create a new column in the sane dataframe with the name "Eligiblity Criteria" where age is > 18, say Eligible else not

```python
if insurance_data.age>18:
    print["Eligible"]
else:
    print["Not Eligible"]
```

```python
def get_age(x):
    if x > 18:
        return "Eligible"
    else:
        return "Not Eligible"

insurance_data["Eligible Criteria"] = insurance_data.age.apply(get_age)
insurance_data 
```

### 7.lambda

#### lambda fuction is ananymous fuction

```python
insurance_data["Eligible Critria"] = insurance_data.age.apply(lambda x:1 if x>20 else 0)
insurance_data
```
