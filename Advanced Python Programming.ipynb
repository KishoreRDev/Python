### Positional Argument

```python
def get_Details(name, place, price):
    print("Client Name = ", name)
    print("Place = ", place)
    print("Price = ", price)

def get_DetailsChennai(name, price,  place= "Chennai"):
    print("Client Name = ", name)
    print("Place = ", place)
    print("Price = ", price)
    


get_Details("Test", 50, "Madurai")    
print()
get_DetailsChennai("Test1", 100, "Theni")
```

### Variable - Lenght Argument

```python
def get_UserDetails(name, phone, *price):
    print(name)
    print(phone)
    print(price)

get_UserDetails("Kishore", 123456789, 60,30,0)
```

## 4.3 Challange to build a user-define functions with loop

```python
def line_by_line(str):
    for i in range(len(str)):
        print(str[i], end="")

def line(str):
    for i in str:
        print(i)

line_by_line("Kishore")

line("Test")
    
```
