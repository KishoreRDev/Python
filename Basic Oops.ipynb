# Real World Analogy 1

```python
# Defining a function

def login(user_data):
    print(f"{user_data["name"]} has logged in" )

def view_task(user_data):
    print(f"{user_data["name"]} is viewing the task" )

def submit_task(user_data):
    if user_data["role"] == "manager":
        print("Manage does not have rights to submit task")
    else:
        print(f"{user_data["name"]} has submitted the task")

def assign_task(user_data):
    if user_data["role"] == "manager":
        print("Manage has assign the task")
    else:
        print("Team Members cannot assign the task")



# Data Collection 

user_data = {"name" : "kishore", "role" : "manager"}

user_data = {"name" : "sowmiya", "role" : "team_member"}

#Function to be called 

login(user_data)
view_task(user_data)
submit_task(user_data)
assign_task(user_data)
```

## With OOPS

Class -A class is a blueprint or template that defines the variables (attributes) and methods (functions) that its objects will have.
An object is an instance of a class. It is a real entity created from the class and can access the class's variables and methods.

```python
class Employee():

    def __init__(self, name):
        self.name = name
        
    def login(self):
        print(f"{self.name} has logged in" )

    def view_task(self):
        print(f"{self.name} is viewing the task" )


class Manager(Employee):
    def assign_task(self):
        print(f"{self.name} has assign the task")

class Team_member(Employee):
    def submit_task(self):
        print(f"{self.name} has submitted the task")

kishore = Team_member("kishore")  
sowmiya = Manager("sowmiya")
print(kishore.name)
kishore.login()
kishore.view_task()
kishore.submit_task()

sowmiya.assign_task()
```
