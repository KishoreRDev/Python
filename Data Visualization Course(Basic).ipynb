# Step 2: Import Dataset

**About Mtcars Data:**

The data was extracted from the 1974 Motor Trend US magazine, and comprises fuel consumption and 10 aspects of automobile design and performance for 32 automobiles (1973--74 models).

```python
mtcars_data = pd.read_csv("mtcars..csv")
mtcars_data
```

#### 3.2 Perform Data visualization for better understanding

```python
### 3.2.1 Univariate Analysis
```

```python
plt.hist(x=mtcars_data.mpg) # speaks about the Distribution of continuous data 
plt.xlabel(xlabel="MPG")
plt.ylabel(ylabel = "Frequency")
plt.title("MPG Distribution")
plt.show()
```

```python
## Insights:
```

```python
## 2 Types of Data Types

1. Discrete - countable and not measurable/ cannot be segregated
2. continuous - Measurable and it has been units / can be segregated
```

### 3.2.2 BIVariate Analysis

```python
mtcars_data.head()
```

```python
plt.bar(x=mtcars_data.hp, height=mtcars_data.model)
plt.show()
```

```python
import seaborn as sns # x axis default discrete data must --> countable and y axis default continuous data must be measurable

sns.barplot(data=mtcars_data, x="cyl", y="mpg")
plt.title("Cylinder Vs MPG")
plt.show()
```
