# Encapsulation

```python
class School:

    def __init__(self):
        self.name = "ABC"
        self.Place = "Chennai"
        self.__Revenue = 500000

    def SchoolRevenue(self):
        print(self.__Revenue)

    def __PrivateMetho(self):
        print(self.__Revenue)

    def publicMetho(self):
        self.__PrivateMetho()

school1 = School()
print(school1.name)
print(school1.Place)
print(school1.SchoolRevenue())
print(school1._School__Revenue)
school1.publicMetho()
        
```

```python
class Person:
    def __init__(self):
        self.__name = None
        self.__age = None

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("please provide proper age")



    @property
    def name(self):
        if self.__name is None:
            print("Please initialize name")
        else:
            return self.__name

    @name.setter
    def name(self, name):
        if len(name) > 20:
            print("Cannot set name having more than 20 characters")
        else:
            self.__name = name


person1 = Person()
person1.name = "Kishore"
print(person1.name)
person1.age = -1
person1.age
```

## Compile time polymorphism is not support in python

```python
class person:

    def __init__(self, name, age):
        self.Name = name
        self.Age = age

    def sleep(self, sleeping_hours):
        print("Sleeping hours is ", sleeping_hours)

    def sleep(self, start, end):
        print("sleeping duration is ", end - start)
        

person1 = person("kishore", 30)
person1.sleep(8)
person1.sleep(8, 6)
```

```python
class person:

    def __init__(self, name, age):
        self.Name = name
        self.Age = age

    def sleep(self, sleeping_hours = None, start = None, end= None):
        if start is not None and end is not None:
            print("Sleeping hours is ", end - start)
        else:
            print("Sleeping houes is ", sleeping_hours)

    

person1 = person("kishore", 30)

person1.sleep(start = 8, end = 6)
person1.sleep(sleeping_hours=10)
```
