# Give 2 examples for each string function

#### 1 . Capitalize() Method

```python
# Example 1
user_name = input("Enter Your Name: ")
print(user_name.capitalize())

# Example 2 
laptop_name = input("Enter Your Laptop Name: ")
print(laptop_name.capitalize())
```

#### 2 . Casefold() Method

```python
# Example 1
user_input = input("Enter Your input: ")
print(user_input.casefold())

# Example 2 
laptop_name = input("Enter Your Laptop Name: ")
print(laptop_name.casefold())
```

#### 3. Center() Method

```python
# Example 1
fruit_name = input("Enter a Fruit Name : ")
print(fruit_name.center(20))

# Example 2
fruit_name = input("Enter a Fruit Name : ")
print(fruit_name.center(20,'*'))
```

#### 4. Count() Method

```python
# Example 1
user_sentence = input("Enter a Sentence : ")
print(user_sentence.count("mango"))

# Example 2
user_sentence = input("Enter a Sentence : ")
print(user_sentence.lower().count("mango", 10, 20))
```

#### 5. Encode() Method

```python
# Example 1
user_txt = input("Enter Your Text : ")
print(user_txt.encode())

# Example 2
user_txt = input("Enter Your Text : ")
print(user_txt.encode(encoding = "ascii", errors = "ignore"))
```

#### 6. Endswith() Method

```python
# Example 1
user_input = input("Enter Your input : ")
print(user_input.endswith("d"))

# Example 2
user_input = input("Enter Your input : ")
print(user_input.endswith("g", 1, 4))

# Example 3
user_input = input("Enter Your input : ")
print(user_input.endswith(("World", "Program")))
```

#### 7. Expandtabs() Method

```python
# Example 1
txt = "H\te\tl\tl\to"
print(txt.expandtabs())

# Example 2
txt = "H\te\tl\tl\to"
print(txt.expandtabs(6))
```

#### 7. Find() Method

```python
# Example 1
txt = "Jupyter Notebook is a good IDE."
print(txt.find("i"))

# Example 2
txt = "Jupyter Notebook is a good IDE."
print(txt.find("i",1, 15))
```

#### 8. Format() Method

```python
# Example 1

jnversion = "Jupyter Notebook Version is {}"
print(jnversion.format(1.0))

# Example 2
user_name = input("Enter Your Name : ")
DOB = input("Enter Your DOB With this format : dd-mm-yyyy : ")
Phonenumber = input("Enter Your Phone Number : ")

print("""
Hi {}, you are successfully registered in this company.
This is your details, please check:

Name = {}
DOB = {}
Phone Number = {}

If anything is missing, you can change your information later.
Thanks for Registration...
""".format(user_name, user_name, DOB, Phonenumber))
```

#### 9. Format_map() Method

```python
# Example 1
txt = "My Laptop Name is {name} and My Windows Version is {version}"

laptop_info = {
    "name": "Acer",
    "version": 11
}

print(txt.format_map(laptop_info))

# Example 2
message = "Hello {user_name}, your account balance is ₹{balance}."

user_details = {
    "user_name": "Kishore",
    "balance": 5000
}

print(message.format_map(user_details))
```

#### 10. Index() Method
