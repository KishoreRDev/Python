#### While Loop

```python
#Example 1
db_password = "Test@123"
attempt = 0
while attempt < 5:
    user_pass =  input("Enter Your Password: ")
    attempt +=1
    if user_pass == db_password:
        print("Your password is matched")
    else:
        print("Please Try again")

#Example 2
count = 5
while count > 0:
    print(count)
    count -=1
```

```python

#Example 1

user_mark = int(input("Enter your mark: "))
if user_mark > 35:
    print("You are pass ")
else:
    print("You are fail")


#Example 2
bank_account_balance = 1000
user_amt = int(input("Enter your amount: "))
if user_amt > bank_account_balance:
    print("Insufficient Balance ")
else:
    print("Transaction Successful")
```

#### if, elif and else

```python
#Example 1

signal_light = input("Enter your signal light: ")
if signal_light.upper() == "GREEN":
    print("Go")
elif signal_light.upper() == "YELLOW":
    print("Get Ready")
else:
    print("Stop")


#Example 2
user_age = int(input("Enter your age: "))
if user_age >= 18:
    print("Your are eligible")
elif user_age < 18:
    print("Your are not eligible")
else:
    print("You are not eligible")
```

#### User Define function without for loop

```python
#Example 1

def reverse_name(name):
    return name.swapcase()

print(reverse_name("Samsung"))

#Example 2
def remove_space(value):
    print(value.strip())

remove_space("       Samsung          ")
```

#### User Define function with for loop

```python

def printname(name):
    for i in name:
        print(i)

printname(input("Enter your name: "))

#Example 2
def print2table(value):
    for i in range(1, value+1):
        print("{} * 2 = {}".format(i, i*2))

print2table(10)
```

#### User Define function with for loop and if condition 

```python

#Example 1

def print_only_str(name):
    for i in name:
        if isinstance(i, str):
            print(i)

print_only_str(["Test", 100, "Man"])

#Example 2
def print_only_even_number(number):
    for i in number:
        if i % 2 == 0:
            print(i)
print_only_even_number([10, 15, 20, 25, 30])
```

#### Break

```python
#Example 1

def find_number(number):
    for i in number:
        if i == 30:
            print("Found the number" , i)
            break

find_number([10, 15, 20, 25, 30])

#Example 2

def find_first_start_with_a(values):
    for i in values:
        if str(i).lower().startswith("a"):
            print(i)
            break
find_first_start_with_a([10, "Ant", "BAT"])
```

#### Continue

```python

#Example 1

def print_odd_numbers(numbers):
    for i in numbers:
        if i % 2 == 0:
            continue
        print(i)

print_odd_numbers([10, 15, 20, 25, 30])

#Example 2

def print_non_empty_strings(words):
    for word in words:
        if str(words) == "":
            continue
        print(word)
print_non_empty_strings(["", 10, "Ant", "BAT", ""])
```

#### for loop with in operator

```python
#Example 1

def find_name(names):

    for name in names:
        if "Samsung" in name:
            print(name)

find_name(["Samsung", "Nokia", "RedMI"])

#Example 2

def print_vowels(txt):
    for t in txt:
        if t in "aeiouAEIOu":
            print(t)
print_vowels("Python Programming")
```

#### for loop with if statement

```python
#Example 1

def find_name(names):

    for name in names:
        if "Samsung" == name:
            print(name)

find_name(["Samsung", "Nokia", "RedMI"])

#Example 2

def print_txt(txt):
    for t in txt:
        if t == "Python Programming":
            print(t)
print_txt("Python Programming")
```

#### for loop with not in operator

```python
#Example 1

def find_name(names):

    for name in names:
        if "Samsung" not in name:
            print(name)

find_name(["Samsung", "Nokia", "RedMI"])

#Example 2

def print_vowels(txt):
    for t in txt:
        if t not in "aeiouAEIOu":
            print(t, end="")
print_vowels("Python Programming")
```
