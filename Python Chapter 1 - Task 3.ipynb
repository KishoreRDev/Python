# Rock, Paper, ScissorsGame

```python
import random as rd

game_list = ["Rock", "Paper", "Scissors"]


def play_game():
    try:
        computer_choice = rd.choice(game_list)
        user_input = input("Enter Rock or Paper or Scissor : ").capitalize()
        print("User Input = {} and Computer Choice = {}".format(user_input, computer_choice))
        if user_input not in  game_list:
            print("Invalid Input!")
        elif user_input == computer_choice:
            print("Match is Tie!")
        elif ((user_input == "Rock" and computer_choice == "Scissors") or (user_input == "Paper" and computer_choice == "Rock")
             or (user_input == "Scissors" and computer_choice == "Paper")):
            print("You Win the Match..")
        else:
            print("Computer Win the Game..")
    except Exception as e:
        print("Error Msg = ", e)
            

while True:
    try:
        play_game()
        result = input("Do you want to play again? (Yes/No): ").capitalize()
        if result != 'Yes':
            print("Thanks for playing!..")
            break
    except Exception as e:
        print("Error Msg = ", e)
```
