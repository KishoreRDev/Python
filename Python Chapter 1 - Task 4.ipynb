# Word Guessing Game

```python
import random as rd

words_list = ["Python", "Java", "NodeJS", "JavaScript"]

computer_choice = rd.choice(words_list)

status = False
max_attempts = 6

print("Guess the Programming Language!")
print(f"You have {max_attempts} attempts.\n")
print("Computer Choice _ _ _ _ _")

for attempt in range(1, max_attempts + 1):
    user_input = input(f"Attempt {attempt}: Enter your guess: ").strip()

    if user_input.lower() == computer_choice.lower():
        print("Congratulations! Your guess is correct.")
        status = True
        break
    else:
        print("Wrong guess. Try again.\n")

if not status:
    print("You've used all your attempts.")
    print(f"The correct answer was: {computer_choice}")
```
