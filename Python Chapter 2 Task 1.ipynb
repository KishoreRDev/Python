# List Examples

## List Properties

1. 1 Dimensional 
2. Mutable
3. A list is an ordered collection
4. A list can contain duplicate values
5. We can access a particular index value and also support slicing.
6. A list can store different data types
7. A list is a heterogeneous

```python

empty_list = []
print(empty_list)
print("Type of empty_list is  = ", type(empty_list))

technologies = ["Python", "SQL", "Linux", "Git", "Docker", "Java"]

print(technologies)

technologies.append("Java")
print("Append : ", technologies)

technologies.extend(["Azure", "Kubernetes"])
print("Extend :", technologies)

technologies.insert(1, ".Net")
print("Insert :", technologies)

technologies.remove("Git")
print("Remove :", technologies)

print(technologies.pop())
print(technologies.pop(4))
print("Pop :", technologies)

print("Index :",technologies.index("SQL"))
print("Index :", technologies.index("Linux", 0, 5))

print("Count :", technologies.count("Java"))

technologies.sort()
print("Sort :", technologies)


technologies.reverse()
print("Reverse :", technologies)

new_list = technologies.copy()
print("New List : ", new_list)

technologies.clear()
print("Old List = {}, New List = {}".format(technologies, new_list))
```

## Set Properties

1. 1 Dimensional 
2. Mutable
3. A set is an unordered collection
4. A set cannot contain duplicate values
5. We cannot access a particular index value and not support slicing.
6. A set can store different data types
7. A set is a heterogeneous

```python
empty_set = set()
print(empty_set)
print("Type of empty_set = ", type(empty_set))

employee_skills = {"Python", "SQL", "Linux", "Git", "Docker"}

employee_skills.add("Java")
print("Add: ", employee_skills)

employee_skills.update({"Azure", "Kubernetes", "Java"})
print("Update: ", employee_skills)

employee_skills.remove("Java")
print("Remove: ", employee_skills)

employee_skills.discard("git")
print("Discard: ", employee_skills)

remove_pop = employee_skills.pop()
print("Remove_pop: ", remove_pop)
print("Pop: ",employee_skills)

employee_skills.clear()
print("Clear: ", employee_skills)

employee_skills = {"Python", "SQL", "Linux", "Git", "Docker"}
new_skills = employee_skills.copy()
print("Copy: ", new_skills)


backend_team = {"Python", "SQL", "Docker", "Git"}

cloud_team = {"AWS", "Docker", "Git", "Linux"}

print("Union: ",backend_team.union(cloud_team))

print("Intersection: ", backend_team.intersection(cloud_team))

print("Difference: ", backend_team.difference(cloud_team))

print("Symmetric_difference: ", backend_team.symmetric_difference(cloud_team))

set_a = {1, 2, 3}
set_b = {2, 3, 4}

set_a.intersection_update(set_b)
print("Intersection_Update: ", set_a)

backend_team = {"Python", "SQL", "Docker", "Git"}

backend_team.difference_update(cloud_team)

print("Difference_Update: ",backend_team, "Cloud Team: ", cloud_team)

backend_team = {"Python", "SQL", "Docker", "Git"}

backend_team.symmetric_difference_update(cloud_team)

print("symmetric_difference_update: ",backend_team, "Cloud Team: ", cloud_team)


small_set = {1, 2}
large_set = {1, 2, 3, 4}
print("issubset 1 :", small_set.issubset(large_set)) 
print("issubset 2 :", small_set <= large_set)        
another_set = {1, 5}
print("issubset 3 :", another_set.issubset(large_set))

main_set = {1, 2, 3, 4, 5}
sub_set = {1, 2, 3}
different_set = {1, 2, 6}
print("issuperset 1:", sub_set.issuperset(main_set))
print("issuperset 2:", main_set.issuperset(sub_set))
print("issuperset 3:", main_set.issuperset(different_set))

A = {1, 2, 3}
B = {4, 5, 6}
c = {1, 2, 4}
fruits = {"apple", "banana", "mango"}
vegetables = {"carrot", "potato"}


print("isdisjoint 1", A.isdisjoint(B))
print("isdisjoint 2", A.isdisjoint(c))
print("isdisjoint 3", fruits.isdisjoint(vegetables))
```

## Tuple Properties

1. 1 Dimensional 
2. IMMutable
3. A tuple is an order collection
4. A tuple can contain duplicate values
5. We can access a particular index value and support slicing.
6. A tuple can store different data types
7. A tuple is a heterogeneous

```python
empty_tuple = ()
print(empty_tuple)
print("Type of empty_tuple = ", type(empty_tuple))

attendance = ("Present", "Absent", "Present", "Present", "Absent")

print("Count :", attendance.count(("present").capitalize()))

students = ("Rahul", "Priya", "Arun", "Kiran")

print("index :", students.index("Arun"))
```

## Dictionary Properties

1. A dictionary is a collection of key-value pairs.
2. A dictionary is mutable (we can add, update, or delete items).
3. A dictionary is an ordered collection.
4. Dictionary keys must be unique, but values can be duplicated.
5. We can access values using their keys.
6. A dictionary can store different data types (heterogeneous).

```python
student = {
    "name": "ram",
    "marks": 90
}

print("Get 1 : ", student.get("address", "not available"))

print("Get 2 : ", student.get("marks", "not available"))

student = {
    "name": "ram",
    "age": 20,
    "city": "Chennai"
}

print("Keys:", student.keys())
print("Values:", student.values())
print("items:", student.items())

student = {
    "name": "ram",
    "age": 20
}
student.update({"age" : 21})
print("update 1:", student)
student.update({"address": "chennai"})
print("update 2:", student)

remove_pop = student.pop("age")
print("pop 1.1: ", remove_pop)
print("pop 1.2: ", student)

remove_pop_item = student.popitem()
print("popitem 1.1:", remove_pop_item)
print("popitem 1.2:", student)

clear_student = student.clear()
print("clear 1.1:", clear_student)
print("clear 1.2:", student)

student = {
    "name": "ram",
    "age": 20
}

new_student = student.copy()
print("new_copy:", new_student, "old_copy:", student)
new_student.update({"address":"chennai"})
print("After Update -- new_copy:", new_student)

student = {
    "name": "ram"
}
student.setdefault("age", 20)
print("setdefault =", student)
```
