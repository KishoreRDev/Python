# Create 3 Dataframe using Dictionaries

## Example 1: Student Marks

```python

import pandas as pd

students = {
    "Student_ID": [101, 102, 103, 104],
    "Name": ["ram", "sanjai", "madhan", "kavin"],
    "Math": [85, 90, 78, 88],
    "Science": [92, 81, 89, 95]
}

student_details = pd.DataFrame(students)
print(student_details)
```

## Example 2: Employee Details

```python
import pandas as pd

employees = {
    "Emp_ID": [1001, 1002, 1003],
    "Name": ["ram", "sam", "kannan"],
    "Department": ["HR", "IT", "Finance"],
    "Salary": [50000, 75000, 68000]
}

employee_details= pd.DataFrame(employees)
print(employee_details)
```

### Example 3: Online Store Orders

```python
import pandas as pd

orders = {
    "Order_ID": [1, 2, 3, 4],
    "Customer": ["Raj", "Priya", "Arun", "Nishant"],
    "Product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "Price": [65000, 25000, 700, 1200]
}

order_detalis = pd.DataFrame(orders)
print(order_detalis)
```
