#### 2. What is the difference between a Jupyter Notebook and Google Colab? 

        Jupyter Notebook and Google Colab are both interactive environments used for writing and running Python code, especially in Data Science and Machine Learning. The main difference is that Jupyter Notebook runs on the local computer, while Google Colab runs on Google's cloud servers. Jupyter Notebook requires Python and the necessary libraries to be installed on your system, whereas Google Colab does not require any installation—you only need a Google account and a web browser. Google Colab also provides free access to GPUs and TPUs, making it suitable for training Machine Learning and Deep Learning models. Jupyter Notebook is preferred when working offline, using local files, or accessing local resources, while Google Colab is ideal for cloud-based development and collaboration.

#### 3. When would you use a list over a set or a tuple? 
    I would use a list when I need an ordered, mutable collection that allows duplicate values. Lists are suitable when I need to add, remove, or modify elements after creation.

    I would use a set when I need only unique elements and want fast membership testing, but the order of elements is not important.

    I would use a tuple when the data should remain unchanged after creation. Tuples are immutable and are commonly used to store fixed data, such as coordinates or configuration values.

#### 4. What is exception handling? Give a real-life example of when it's useful.

        Exception handling is a mechanism used to handle runtime errors without terminating the program. It allows us to detect errors and continue executing the program gracefully. In Python, exception handling is implemented using try, except, else, and finally blocks. For example, in an ATM system, if a user enters an incorrect PIN or tries to withdraw more money than the available balance, the system should display an error message instead of crashing. Exception handling ensures the program handles such situations gracefully and allows the user to try again.

#### 5. Define DataFrame and Series in Pandas. What’s the difference? 

    A Series is a one-dimensional data structure in Pandas that stores a single column of data. It has an index and can store values such as numbers, strings, or dates.

    A DataFrame is a two-dimensional data structure in Pandas that stores data in rows and columns, similar to an Excel sheet or a database table. Each column in a DataFrame is a Series.

    The main difference is that a Series contains only one column of data, whereas a DataFrame contains multiple columns of data.

#### 7.How do NumPy and Pandas differ in terms of functionality? 
    NumPy and Pandas are both Python libraries used in Data Science, but they have different purposes.

    NumPy is mainly used for numerical computations. It works with arrays and provides fast mathematical operations such as addition, subtraction, multiplication, matrix operations, and statistical calculations.

    Pandas is mainly used for data manipulation and analysis. It works with Series and DataFrames, making it easy to read, clean, filter, merge, and analyze data from files like CSV and Excel.

#### 8. What is the role of libraries like Seaborn and Matplotlib in a data project? 
    Seaborn and Matplotlib are Python libraries used for data visualization. They help us convert data into charts and graphs, making it easier to understand patterns, trends, and relationships in the data.
    
    Matplotlib is used to create basic charts such as line charts, bar charts, pie charts, scatter plots, and histograms.

    Seaborn is built on top of Matplotlib and is used to create more attractive and advanced statistical visualizations with less code.

#### 9.When should you use .loc[] and when .iloc[]? 
    Both .loc[] and .iloc[] are used to select data from a Pandas DataFrame.

    .loc[] is used to select rows and columns using their labels (names).

    .iloc[] is used to select rows and columns using their integer positions (index numbers).
