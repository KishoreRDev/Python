#### 1.Write a program to take a name and age from the user and print a greeting message. 

```python


user_name = input("Enter Your Name").strip()
user_age = input("Enter your Age").strip()
if not user_name:
    print("Name cannot be empty.")
elif not user_age:
    print("Age cannot be empty.")
elif not user_age.isdigit():
    print("Please enter a valid age.")
else:
    print("Welcome to Python Programming.")
```

#### 2.Create a list of numbers and print only even numbers using a for loop. 

```python
list_of_number = [2,4,5,7,10,15,6]
even_list = []
for i in list_of_number:
    if i%2==0:
        even_list.append(i)
print("This is an even numbers = ", even_list)
```

#### 3.Write a program that handles division by zero using try-except. 

```python
try:
    user1_input = input("Enter a number")
    user2_input = input("Enter a number")
    if not user1_input or not user2_input:
        print("Please enter a number")
    elif not user1_input.isdigit() or not user2_input.isdigit():
        print("Please enter a valid number")
    else:
        result = int(user1_input)/int(user2_input)
        print("Result is", result)
except Exception as e:
    print("Cannot divide by zero", e)
```

#### 4.Given a string, count how many vowels it has. 

```python
user_name = input("Enter Your Name").strip()
vowels = "aeiouAEIOU"
count = 0
for i in user_name:
    if i in vowels:
        print(i, end=" ")
        count +=1
print("\nTotal number of vowels: ", count)
```
