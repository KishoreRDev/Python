## Merge Function 1 Example 

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Marks": [85, 90, 78, 88]
})

result = pd.merge(students, marks, on="Student_ID", how="inner")
result
```

```python
result = pd.merge(students, marks, on="Student_ID", how="left")
result
```

```python
result = pd.merge(students, marks, on="Student_ID", how="right")
result
```

```python
result = pd.merge(students, marks, on="Student_ID", how="outer")
result
```

## List Comprehension 2 examples 

```python
# Example 1 salary greater than 50,000

employees = [
    {"name": "Raj", "salary": 45000},
    {"name": "Sam", "salary": 60000},
    {"name": "Ram", "salary": 75000},
    {"name": "Hari", "salary": 40000}
]

emp_name = [emp["name"] for emp in employees if emp["salary"] > 50000]
emp_name
```

```python
# Example 2
list1 = [1, 2, 3]
list2 = [10, 20]

result = [(i, j) for i in list1 for j in list2 if j>10]
result
```

# Task 4

## Lambda Function 2 examples with Map, Filter, Reduce

### Map

```python
# Example 1: Square every number
numbers = [1,2, 3, 4, 5]

result = list(map(lambda x: x**2, numbers))
result
```

```python
# Example 2: Convert names to uppercase
names = ["raj", "sam", "ram"]

result = list(map(lambda x: x.upper(), names))
result
```

### Filter

```python
# Example 1: Get even numbers
numbers = [1,2,3,4,5,6,7,8]

result = list(filter(lambda x: x%2==0, numbers))
print(result)
```

```python
# Example 2: Students with marks >= 50

marks = [45, 67, 30, 89, 55]
result = list(filter(lambda x: x>=50, marks))
print(result)
```

### Reduce

```python
# Example 1: Sum of numbers

from functools import reduce
numbers = [1,2,3,4,5]
result = reduce(lambda x,y: x+y, numbers)
result
```

```python
# Example 2: Find the largest number
from functools import reduce

numbers = [12, 8, 45, 22, 30]
result = reduce(lambda x, y: x if x>y else y, numbers)
result
```
